```python
import subprocess; subprocess.run(['ffmpeg','-version'], capture_output=True); print('✓ FFmpeg OK')
```

```python
from pathlib import Path; OUTPUT = Path('/content/output'); OUTPUT.mkdir(exist_ok=True); print('✓ Pasta:', OUTPUT)
```

```python
import subprocess; VIDEO = OUTPUT / 'relaxing.mp4'; cmd = [
  'ffmpeg','-y',
  '-f','lavfi','-i','color=c=#0a1a2e:s=1280x720:d=5:rate=30',
  '-f','lavfi','-i','anoise=amount=0.02:color=brown:duration=5',
  '-vf','drawtext=fontsize=36:fontcolor=white:text=Relaxing Music:x=(w-text_w)/2:y=h/2-40,drawtext=fontsize=24:fontcolor=#aaa:text=Stress Relief Meditation:x=(w-text_w)/2:y=h/2+20',
  '-c:v','libx264','-preset','ultrafast','-crf','30',
  '-c:a','aac','-b:a','96k','-pix_fmt','yuv420p','-shortest',
  str(VIDEO)
]
result = subprocess.run(cmd, capture_output=True, text=True)
if VIDEO.exists():
  print(f'✓ Video created: {VIDEO.stat().st_size/1024:.1f} KB')
else:
  print('✗ Failed'); print(result.stderr[-300:])
```
