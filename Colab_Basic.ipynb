```python
import subprocess, os
from pathlib import Path

# Cria pasta
os.makedirs('/content/out', exist_ok=True)
output = '/content/out/video.mp4'

# Comando simples
cmd = 'ffmpeg -y -f lavfi -i color=c=blue:s=640x480:d=3 -pix_fmt yuv420p ' + output
print('Executando:', cmd)
os.system(cmd)

# Verifica
if os.path.exists(output):
    size = os.path.getsize(output)
    print(f'OK! Arquivo: {size} bytes')
else:
    print('ERRO - arquivo nao criado')
```
